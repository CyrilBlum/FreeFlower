# Zufall simulieren

**Einstiegsfrage:** Ein Casino verliert an rund jedem dritten Abend gegen eine Person, die stur auf Rot setzt. Weshalb gehört es trotzdem nie dem Pech?

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2027)


def pruefe(name, wert, soll, toleranz):
    """Selbstkontrolle: vergleicht Ihren Wert mit dem erwarteten Bereich."""
    if wert is None:
        print(f"{name}: noch nicht bearbeitet")
    elif abs(wert - soll) <= toleranz:
        print(f"{name}: passt ({wert:.4g})")
    else:
        print(f"{name}: prüfen Sie Ihren Code, erwartet wird etwa {soll:.4g}, Sie haben {wert:.4g}")
```

## Aufgabe 1: Roulette an einem Abend
Eine Person setzt 100-mal je 1 CHF auf Rot (18 von 37 Feldern). Simulieren Sie 200'000 Abende. Speichern Sie in `gewinn` den Gewinn jedes Abends.

*Hilfestellung:* `rng.binomial(100, p, size=n)` liefert die Anzahl Treffer. Gewinn = Treffer minus Fehlschläge.

```python
n = 200_000
gewinn = None
gewinn = 2 * rng.binomial(100, 18 / 37, size=n) - 100
anteil_plus = None if gewinn is None else np.mean(gewinn > 0)
pruefe('Anteil Abende im Plus', anteil_plus, 0.355, 0.01)
```

## Aufgabe 2: Verteilung ansehen
Zeichnen Sie ein Histogramm von `gewinn` und markieren Sie den Durchschnitt mit `plt.axvline`.

```python
plt.hist(gewinn, bins=40)
plt.axvline(gewinn.mean(), color='red')
plt.xlabel('Gewinn in CHF')
plt.show()
```

## Aufgabe 3: Das Casino zieht Bilanz
An einem Abend spielen 500 Personen dieses Programm. Simulieren Sie 10'000 Abende und bestimmen Sie den Anteil Abende, an denen das Casino insgesamt Geld verliert.

*Hilfestellung:* Summe von 500 Abenden = 50'000 Spiele. Sie können direkt `rng.binomial(50_000, p, size=...)` verwenden.

```python
verlust_casino = None
verlust_casino = np.mean(2 * rng.binomial(50_000, 18 / 37, size=10_000) - 50_000 > 0)
pruefe('Anteil Verlustabende Casino', verlust_casino, 0.0, 0.005)
```

## Challenge: Wie viele Simulationen braucht es?
Wie viele Würfelwürfe brauchen Sie, damit der Durchschnitt mit 95 % Sicherheit höchstens 0.01 von 3.5 abweicht?

*Hilfestellung:* Der Standardfehler ist $\sigma/\sqrt{n}$, die 95-%-Grenze liegt bei 1.96 Standardfehlern. Für einen Würfel gilt $\sigma^2 = 35/12$.

```python
n_noetig = None
n_noetig = (1.96 * np.sqrt(35 / 12) / 0.01) ** 2
pruefe('Anzahl Würfe', n_noetig, 112_047, 500)
```
